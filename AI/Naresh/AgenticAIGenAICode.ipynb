```python
# Mini Project - The Smart Keyword Extractor

# Scenario:
# You are analyzing customer complaints.
# The complaints are messy.
# You want to extract the Core Issue (Keywords)
# so you can tag them in a database.

# Tasks:
# 1. Input: "My batteries aren't charging properly and the screen is cracked"
# 2. Clean the stop words : ('My', 'and', 'the', 'is')
# 3. Lemmatize the rest: ('batteries -> battery', 'cracked -> crack')
# 4. Output -> ['battery', 'charge', 'properly', 'screen', 'crack']
```

```python
import spacy
nlp = spacy.load('en_core_web_md')

# Reusable block of code which I can use whenever I want -> Function are blocks of code
def extract_keywords(text):
    result = nlp(text)

    keywords = []

    for token in result:
        # Filter out stop words
        if not token.is_stop and not token.is_punct:

            # Lemmatisation
            root_word = token.lemma_

            keywords.append(root_word)

    return keywords
```

```python
extract_keywords("My batteries aren't charging properly and the screen is cracked")
```

```python
import torch
from transformers import pipeline

pipe = pipeline(
    "text-generation",
    model = "TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    torch_dtype = torch.float16,
    device_map = 'cuda'
)
```

```python
# System Prompt
# User Prompt
# Assistant
# Free model do not have separate roles, we need to specify everything in one string
# <|system|>
# <|user|>
# <|assistant|>
```

```python
def query_local_llm(user_input, system_instructions):

    prompt = f'<|system|>{system_instructions}  <|user|>{user_input}   <|assistant|>'

    # Run the model
    outputs = pipe(
        prompt,
        max_new_tokens=256, # Limit the length of the answer
        do_sample = True, # Add a little creativity
        temperature = 0.7, # Creativity Level
        top_k = 50, # Only choose from top 50 words
        top_p = 0.95 # Minimum the probability should be 95%
    )

    full_text = outputs[0]['generated_text']

    return full_text
```

```python
result = query_local_llm('What is the capital of France?', 'You are a helpful assistant')
result.split("<|assistant|>")[-1].strip()
```

```python
import re
raw_chunks = re.split(r"\n## ", raw_handbook_text) # \n -> newline character, means 'enter'
for chunk in raw_chunks:
    print(chunk)
    print("------------------------------------------------")
```

```python
# Storing our chunks into the vector database
chunk_ids = [f'{i}' for i in range(len(raw_chunks))]
```

```python
def hr_bot(user_question):
    print(f"\n👤 User: {user_question}")

    # Phase 1: Retrieval of the relevant data
    result = collection.query(
        query_texts = [user_question],
        n_results = 1 # only the top matching result
    )

    # Extract the sentence from the result
    retrieved_context = result['documents'][0][0]

    print(f"🔍 System Retrieved: '...{retrieved_context}")

    # Phase 2: Generation
    # Construct a highly restrictive system prompt
    system_prompt = f"""
    You are an HR Assistant.
    You must answer the user's question using ONLY the provided Context below.
    If the answer is not in the context, say "I don't know based on the handbook."
    Do NOT invent information.

    CONTEXT:
    {retrieved_context}
    """

    # Call Groq (Llama-3)
    completion = groq.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_question}
        ],
        temperature=0.0 # Strict and factual!
    )

    answer = completion.choices[0].message.content
    print(f"🤖 HR Bot: {answer}\n" + "-"*40)
```
