```python
# RUN THIS CELL FIRST TO INITIALIZE THE GRADER
class AutoGrader:
    def __init__(self):
        self.total_score = 0
        self.max_score = 50

    def check(self, q_num, condition, points=10):
        try:
            if condition():
                print(f"✅ Question {q_num}: Correct! (+{points} points)")
                self.total_score += points
            else:
                print(f"❌ Question {q_num}: Incorrect. Try again!")
        except Exception as e:
            print(f"⚠️ Question {q_num}: Error in your code -> {e}")

    def show_score(self):
        print("="*30)
        print(f"🏆 FINAL SCORE: {self.total_score} / {self.max_score}")
        if self.total_score == self.max_score:
            print("🌟 PERFECT SCORE! You are ready for Agentic AI!")
        print("="*30)

grader = AutoGrader()
print("Grader initialized! You may begin.")
```
