# 🤖 OrderBot Challenge — Graded Test

You're building **OrderBot**, an AI food-delivery assistant. Each task tests one coding concept from the lecture. Write your code, run the ✅ CHECK cell, and watch your rank climb.

| Rank | Score |
|---|---|
| 🍕 Trainee Chef | 0 – 25% |
| 🛵 Delivery Rider | 25 – 50% |
| 📋 Order Manager | 50 – 70% |
| 🤖 Bot Architect | 70 – 85% |
| 🧠 Senior AI Engineer | 85 – 99% |
| 👑 OrderBot Master | 100% |

**11 tasks · 100 points total**


## ⚙️ Setup — run this first

````python
class Grader:
    RANKS = [
        (0.00, "🍕 Trainee Chef",         "The kitchen just hired you."),
        (0.25, "🛵 Delivery Rider",        "You know the routes now."),
        (0.50, "📋 Order Manager",         "Running the floor smoothly."),
        (0.70, "🤖 Bot Architect",         "Building bots like a pro."),
        (0.85, "🧠 Senior AI Engineer",    "Machines do your bidding."),
        (1.00, "👑 OrderBot Master",       "The bot was your idea all along."),
    ]
    def __init__(self, total=100):
        self.scores, self.max_pts = {}, {}
        self.total = total

    @property
    def earned(self): return sum(self.scores.values())

    def _bar(self, pct, w=28):
        f = int(round(pct * w))
        return " " * f + " " * (w - f)

    def _rank(self):
        pct = self.earned / self.total
        r = self.RANKS[0]
        for t, n, b in self.RANKS:
            if pct >= t: r = (t, n, b)
        return r

    def check(self, tid, passed, pts, ok_msg="", fail_msg=""):
        self.max_pts[tid] = pts
        already = self.scores.get(tid, 0) == pts
        if passed:
            self.scores[tid] = pts
            print(("🔁 Already complete" if already else f"✅ +{pts} pts!") + f"  [Task {tid}]")
            if ok_msg and not already: print(f"   {ok_msg}")
        else:
            if not already: self.scores[tid] = 0
            print(f"❌ Task {tid} not passing yet.  (0 / {pts} pts)")
            if fail_msg: print(f"   {fail_msg}")
        pct = self.earned / self.total
        _, n, b = self._rank()
        print()
        print(f"   {self.earned}/{self.total}  [{self._bar(pct)}]  {pct*100:.0f}%")
        print(f"   {n} - {b}")
        print()

    def scorecard(self):
        pct = self.earned / self.total
        _, n, b = self._rank()
        print("=" * 52)
        print("  🏆  FINAL SCORECARD")
        print("=" * 52)
        for tid in sorted(self.max_pts):
            e, t = self.scores.get(tid, 0), self.max_pts[tid]
            print(f"  {'✅' if e==t else '❌'}  Task {tid:<5s} {e:>3d}/{t}")
        print(f"  {'─'*42}")
        print(f"  TOTAL  {self.earned:>3d}/100  [{self._bar(pct)}]  {pct*100:.0f}%")
        print(f"  {n}")
        print(f"  {b}")
        print("=" * 52)

grader = Grader()
print("Scoreboard ready. Good luck! 🍕")
````
